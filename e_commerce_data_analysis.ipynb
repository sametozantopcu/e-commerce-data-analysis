```python
### 10. Categorical String Normalization and Cardinality Validation on `sehir`
df["sehir"].unique()
```

```python
len(df["sehir"].unique())
```

```python
sehir_modu = df['sehir'].mode()[0]
df['sehir'] = df['sehir'].fillna(sehir_modu)
sehir_modu
# During cardinality inspection of the `sehir` feature, null entries (`NaN`) were identified. To ensure downstream string operations execute cleanly and categorical integrity is maintained, missing records were imputed with the feature's statistical mode, `"İstanbul"`.
```

```python
df["sehir"].unique()
```

```python
def metin_normalize(series):

  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇", "i", regex=False)
      .str.replace("ı", "i", regex=False)
      .str.replace("ö", "o", regex=False)
      .str.replace("ü", "u", regex=False)
      .str.replace("ç", "c", regex=False)
      .str.replace("ş", "s", regex=False)
      .str.replace("ğ", "g", regex=False)
  )
```

```python
df["sehir"] = metin_normalize(df["sehir"])
df["sehir"].unique()
```

```python
duzeltme_haritasi_sehir = {
    "istanbul": "İstanbul",
    "ankara": "Ankara",
    "izmir": "İzmir",
    "antalya": "Antalya",
    "konya": "Konya",
    "bursa": "Bursa",
    "adana": "Adana",
    "gaziantep": "Gaziantep",
    "mersin": "Mersin",
    "kayseri": "Kayseri",
    "eskisehir": "Eskişehir",
    "samsun": "Samsun",
    "trabzon": "Trabzon",
    "kocaeli": "Kocaeli"
}
df["sehir"] = df["sehir"].replace(duzeltme_haritasi_sehir)
df["sehir"].unique()
```

```python
len(df["sehir"].unique())
```

#### Following the imputation of the single missing (`NaN`) entry using the categorical mode (`"İstanbul"`), a custom normalization function paired with a mapping dictionary was applied to resolve casing discrepancies and Turkish character artifacts, enforcing proper title-casing across all city names. As a result of eliminating typographical inconsistencies, the unique category count (cardinality) decreased from 20 to 14, establishing complete categorical standardization across the feature.

### 15. General Evaluation & Business Insights

**1. Distribution Analysis & Demographics**

* **Payment Preferences (Count Plot):** *Kredi Kartı* (Credit Card) heavily dominates as the primary payment method, whereas *Kapıda Ödeme* (Cash on Delivery) is the least preferred option.

**2. Descriptive Statistics Overview**
* **Pricing & Outliers:** The extreme maximum values (up to 63,870 TL) in both `birim_fiyat` and `toplam_tutar`—which sit significantly above their respective means—clearly indicate the presence of premium/luxury spending within the dataset.
* **Customer Satisfaction:** The overall mean for `musteri_puani` stands at a solid 3.99 out of 5, reflecting a generally high level of customer satisfaction.

**3. Strategic Answers & Actionable Recommendations**

  * *Recommendation:* Specialized campaigns or targeted discounts can be deployed to further capitalize on the strong sales momentum in Electronics. Additionally, cross-promotional strategies could be designed to boost the revenue contribution of other product categories.

However, the *VIP* segment slightly trails in both population size and average rating. Enhancing the premium service quality and offering exclusive perks could elevate their satisfaction, subsequently growing the overall VIP customer base.
